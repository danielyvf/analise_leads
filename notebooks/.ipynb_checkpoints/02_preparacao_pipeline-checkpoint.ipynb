```python
#Ler o dataset original
base_leads = pd.read_csv(r"C:\Users\dvent\OneDrive\Documentos\pipeline_sdr\data_sdr\leads.csv")
```

```python
#Ajustar tipos
base_leads["lead_id"] = base_leads["lead_id"].astype(int)
base_leads["empresa_id"] = base_leads["empresa_id"].astype(int)
base_leads["sdr_owner"] = base_leads["sdr_owner"].astype(int)
base_leads["criado_em"] = pd.to_datetime(base_leads["criado_em"])
```

```python
#Agregações para o funil
#Leads por SDR
leads_por_sdr = base_leads.groupby("sdr_owner")["lead_id"].count().reset_index()
leads_por_sdr.rename(columns={"lead_id": "total_leads"}, inplace=True)
```

```python
#Leads por origem
leads_por_origem = base_leads["origem"].value_counts().reset_index()
leads_por_origem.columns = ["origem", "total_leads"]
```

```python
#Leads por cargo
leads_por_cargo = base_leads["cargo"].value_counts().reset_index()
leads_por_cargo.columns = ["cargo", "total_leads"]
```

```python
#Leads por mês
base_leads['mes_criacao'] = base_leads['criado_em'].dt.to_period('M')
leads_por_mes = base_leads.groupby('mes_criacao')['lead_id'].count().reset_index()
leads_por_mes.rename(columns={"lead_id": "total_leads"}, inplace=True)
```

```python
# Definindo o diretório onde os CSVs serão salvos
data_sdr = r"C:\Users\dvent\OneDrive\Documentos\pipeline_sdr\data_sdr"  # Caminho absoluto

# Verificar se o diretório 'data_sdr' existe, e se não, criar a pasta
if not os.path.exists(data_sdr):
    os.makedirs(data_sdr)

# Supondo que você tenha os DataFrames 'leads_por_sdr', 'leads_por_origem', 'leads_por_cargo' e 'leads_por_mes' gerados
# Aqui estamos apenas salvando esses DataFrames como arquivos CSV dentro da pasta 'data_sdr'

# Salvar os CSVs finais na pasta 'data_sdr'
leads_por_sdr.to_csv(os.path.join(data_sdr, "leads_por_sdr.csv"), index=False)
leads_por_origem.to_csv(os.path.join(data_sdr, "leads_por_origem.csv"), index=False)
leads_por_cargo.to_csv(os.path.join(data_sdr, "leads_por_cargo.csv"), index=False)
leads_por_mes.to_csv(os.path.join(data_sdr, "leads_por_mes.csv"), index=False)

# Imprimir mensagem confirmando que os arquivos foram gerados
print("CSV finais gerados na pasta:", data_sdr)

```
