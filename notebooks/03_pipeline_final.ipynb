```python
import pandas as pd
import os

# Pasta onde estão os CSVs finais
data_sdr = r"C:\Users\dvent\OneDrive\Documentos\pipeline_sdr\data_sdr"

# Lista de CSVs finais
csvs = ["leads_por_sdr.csv", "leads_por_origem.csv", "leads_por_cargo.csv", "leads_por_mes.csv"]

# Criar um DataFrame resumido
resumo = []

for arquivo in csvs:
    caminho_arquivo = os.path.join(data_sdr, arquivo)
    
    if os.path.exists(caminho_arquivo):
        df = pd.read_csv(caminho_arquivo)
        colunas = ", ".join(df.columns)
        linhas = df.shape[0]
        resumo.append({"arquivo": arquivo, "colunas": colunas, "linhas": linhas})
    else:
        resumo.append({"arquivo": arquivo, "colunas": "Arquivo não encontrado", "linhas": 0})

# Mostrar a tabela resumida
resumo_df = pd.DataFrame(resumo)
resumo_df
```
